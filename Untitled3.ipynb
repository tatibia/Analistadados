```python
# Entrada de dados
potencia_lampada = float(input("Digite a potência da lâmpada (em Watts): "))
largura = float(input("Digite a largura do cômodo (m): "))
comprimento = float(input("Digite o comprimento do cômodo (m): "))

# Cálculos
area = largura * comprimento
potencia_total = area * 3  # 3 Watts por m²[cite: 1]
qtd_lampadas = potencia_total / potencia_lampada

print(f"Área do cômodo: {area:.2f} m²")
print(f"Quantidade de lâmpadas necessárias: {qtd_lampadas:.1f}")
```
