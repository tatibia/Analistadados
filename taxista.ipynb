```python
# Entrada de dados
km_inicio = float(input("Marcação do odômetro no início do dia (km): "))
km_fim = float(input("Marcação do odômetro no final do dia (km): "))
litros = float(input("Quantidade de combustível gasto (L): "))
recebido = float(input("Valor total recebido dos passageiros (R$): "))

# Cálculos
distancia = km_fim - km_inicio
media_consumo = distancia / litros
custo_combustivel = litros * 6.15  # R$ 6,15 por litro[cite: 1]
lucro_liquido = recebido - custo_combustivel

print(f"Média de consumo: {media_consumo:.2f} km/L")
print(f"Lucro líquido do dia: R$ {lucro_liquido:.2f}")
```
