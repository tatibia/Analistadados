```python
for i in range(1, 11):
    print(f"\n--- ALUNO {i} ---")
    nota1 = float(input("Digite a 1ª nota: "))
    nota2 = float(input("Digite a 2ª nota: "))
    optativa = float(input("Digite a nota optativa (-1 se não fez): "))

    if optativa != -1:
        if nota1 < nota2:
            nota1 = max(nota1, optativa)
        else:
            nota2 = max(nota2, optativa)

    media = (nota1 + nota2) / 2
    print(f"Média do aluno {i}: {media:.2f}")

    if media >= 6.0:
        print("Status: Aprovado")
    elif media >= 3.0:
        print("Status: Recuperação")
    else:
        print("Status: Reprovado")
```
