```python
codigo = int(input("Digite o código de origem do produto: "))

match codigo:
    case 1:
        print("Região: Sul")
    case 2:
        print("Região: Norte")
    case 3:
        print("Região: Leste")
    case 4:
        print("Região: Oeste")
    case 5 | 6:
        print("Região: Nordeste")
    case 7 | 8 | 9:
        print("Região: Sudeste")
    case 10:
        print("Região: Centro-Oeste")
    case 11:
        print("Região: Noroeste")
    case _:
        print("Região: Importado")
```
