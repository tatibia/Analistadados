```python
resultados = []

for i in range(1, 6):
    print(f"\n--- ALUNO {i} ---")
    nota1 = float(input("Digite a 1ª nota: "))
    nota2 = float(input("Digite a 2ª nota: "))
    optativa = float(input("Digite a nota optativa (-1 se não fez): "))

    # Substitui a menor nota se a optativa foi realizada
    if optativa != -1:
        if nota1 < nota2:
            nota1 = max(nota1, optativa)
        else:
            nota2 = max(nota2, optativa)

    media = (nota1 + nota2) / 2

    # Verifica situação
    if media >= 6.0:
        status = "Aprovado"
    elif media >= 3.0:
        status = "Recuperação"
    else:
        status = "Reprovado"

    # Salva o resultado formatado na lista
    resultados.append(f"Aluno {i} - Média: {media:.2f} ({status})")

# Exibe o resumo final acumulado
print("\n=== RESUMO FINAL DOS ALUNOS ===")
for res in resultados:
    print(res)
```
